### Exportação

```python
# Se lista_pav contiver apenas cotas numéricas, converte para sólidos antes de exportar
if lista_pav and isinstance(lista_pav[0], (int, float)):
    lista_pav_solids = [b3d.Box(30, 40, pap).translate((0, 0, cota)) for cota in lista_pav]
else:
    lista_pav_solids = lista_pav

# Atribui um label (nome) a cada pavimento e agrupa tudo em um Compound
for i, pav in enumerate(lista_pav_solids):
    pav.label = f"pav_{i}"

assy = b3d.Compound(children=lista_pav_solids)

# Exporta o Compound para um arquivo STEP
b3d.export_step(assy, "output.step")

# Alternativa: unir (fundir) os sólidos com "+" antes de exportar,
# equivalente ao mode="fused" do cq.Assembly.export() do CadQuery
# fused = lista_pav[0]
# for pav in lista_pav[1:]:
#     fused = fused + pav
# export_step(fused, "output.step")
```
