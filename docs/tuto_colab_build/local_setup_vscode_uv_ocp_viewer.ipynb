# Setup Local: VS Code + uv + OCP CAD Viewer

Este notebook mostra um fluxo completo para executar os notebooks localmente com ambiente reproduzivel para alunos.

## Objetivos

Ao final, voce tera:
- VS Code instalado
- uv instalado
- Ambiente local criado com Python 3.12
- Kernel Jupyter selecionado no VS Code
- Visualizacao 3D via OCP CAD Viewer

## 1) Instalar VS Code

Site oficial: https://code.visualstudio.com/

Durante a instalacao no Windows, marque:
- Add to PATH
- Register code as an editor

Depois abra esta pasta do projeto no VS Code.

## 2) Instalar uv

### Windows (PowerShell)
```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

### macOS / Linux
```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

Verificacao:
```bash
uv --version
```

## 3) Criar ambiente local e instalar dependencias

No terminal na raiz do repositorio:
```bash
uv python pin 3.12
uv venv --seed
uv sync --dev --group notebook --group ifc
```

Observacao:
- `--seed` garante `pip` no ambiente
- `--group notebook` instala stack de notebook
- `--group ifc` instala stack IFC

## 4) Selecionar kernel no VS Code

1. Abra um notebook `.ipynb`.
2. Clique no seletor de kernel (canto superior direito).
3. Escolha `Python Environments` e selecione `.venv`.
4. Se necessario: `Ctrl+Shift+P` -> `Developer: Reload Window`.

Se o kernel sumir da lista, normalmente e ambiente sem `pip`/`ipykernel` ou Python fora da versao esperada.

## 5) Extensoes recomendadas no VS Code

Instale estas extensoes:
- `ms-python.python` (Python)
- `ms-toolsai.jupyter` (Jupyter)
- `ms-python.vscode-pylance` (Pylance)
- `ms-python.debugpy` (Python Debugger)
- `bernhard-42.ocp-cad-viewer` (OCP CAD Viewer)
- `yzhang.markdown-all-in-one` (opcional, docs)
- `esbenp.prettier-vscode` (opcional, markdown/json)

## 6) Instalar pacote Python para OCP CAD Viewer

A extensao do VS Code trabalha junto com o pacote Python `ocp_vscode`.

No terminal da raiz:
```bash
uv add ocp-vscode
```

Ou, se preferir manter em grupo dev/notebook, ajuste o `pyproject.toml` e rode `uv sync`.

## 7) Trocar `show` para OCP CAD Viewer

### Antes (simple viewer)
```python
import build123d as b3d
from cadquery_simple_viewer import show
```

### Depois (OCP CAD Viewer)
```python
import build123d as b3d
from ocp_vscode import show
```

Uso basico permanece simples:
```python
box = b3d.Box(80, 60, 40)
show(box)
```

````python
import build123d as b3d
from ocp_vscode import show

caixa = b3d.Box(100, 80, 40)
cilindro = b3d.Cylinder(radius=20, height=80).translate((80, 0, 0))

show([caixa, cilindro], names=["Caixa", "Cilindro"])
````

## 8) Troubleshooting rapido

### Kernel aparece e some
- Confirme Python 3.12 no projeto (`uv python pin 3.12`)
- Recrie ambiente: `uv venv --seed`
- Reinstale deps: `uv sync --dev --group notebook --group ifc`

### `ModuleNotFoundError` para viewer
- Para notebooks em `docs/tuto_colab_build`, existe shim local para `cadquery_simple_viewer`
- Para OCP, confirme `uv add ocp-vscode`

### Mensagem `'name' table stringOffset incorrect...`
- Geralmente e warning de fonte na stack grafica
- Se a modelagem e visualizacao funcionam, pode ignorar
